Для анализа мы выбрали 2 файла, посвященных теме туризма, поскольку нам стало интересно проанализировать, как сильно коронавирусные ограничения повлияли на отдых россиян за границей. В 1-ом файле содержатся данные из опроса ВЦИОМ "Как я провел это лето - 2023", во 2-ом - данные Росстата о числе выездных туристских поездок из России в 2019 и в 2023 годах. Данные из этих 2 файлов позволят выяснить, как поменялись предпочтения россиян относительно летнего отдыха на протяжении 5 лет. 

```python
# обработка 1-ого файла

result_d = {} # словарь со всеми данными {ответ 1: [данные за 2019, данные за 2020 ...], ответ 2: [данные за 2019, данные за 2020 ...] ...}

with open ('tourism1.csv', mode = 'r', encoding = 'utf-8') as infile1:
    for line in infile1:
        lines = line.strip()
        answer, percent19, percent20, percent21, percent22, percent23 = lines.split(',')
        percent19 = int(percent19) # преобразовываем проценты 2019 года в числа
        percent20 = int(percent20) # преобразовываем проценты 2020 года в числа
        percent21 = int(percent21) # преобразовываем проценты 2021 года в числа
        percent22 = int(percent22) # преобразовываем проценты 2022 года в числа
        percent23 = int(percent23) # преобразовываем проценты 2023 года в числа
        # заполняем словарь result_d:
        if not answer.endswith('?'):
          result_d[answer] = [percent19, percent20, percent21, percent22, percent23]
```

После обработки данных файла мы перешли к их анализу с помощью описательных статистик:

```python
# ищем среднее значение по ответу "За границей/в другой стране"
aver = sum(result_d['За границей/в другой стране']) / len(result_d['За границей/в другой стране'])
```

В среднем за 5 лет на вопрос о том, где и как вы провели лето, 5,2% россиян отвечали, что были за границей. Это значение невелико, что можно объяснить тем, что на протяжении 5 рассмотренных лет вводились ковидные ограничения, санкции, связанные с началом специальной военной операции, а также в целом реальные доходы населения снизились, а курсы иностранной валюты, наоборот, укрепились, что привело к подорожанию путевок за границу.

```python
# находим позицию, которую занимает ответ "За границей/в другой стране" в 2019 году
top_ans_19 = [] # рейтинг ответов за 2019 год
for k, v in sorted(result_d.items(), key = lambda x: x[1][0], reverse = True):
    top_ans_19.append(k)
place_19 = top_ans_19.index('За границей/в другой стране') + 1 # позиция, которую занимает ответ "За границей/в другой стране" в 2019 году
```

Полученное значение можно интерпретировать следующим образом: ответ "За границей, в другой стране" в 2019 году занимает 5 (из 8) место среди других ответов. Чаще опрошенные отвечали, что провели отпуск дома, занимаясь своими делами; на даче; не брали отпуск вообще или посетили другой город России. Из этого наблюдения следует, что поездки за границу не являются распространенным вариантов отпуска у россиян, что подтверждается и средним значением.

```python
# находим позицию, которую занимает ответ "За границей/в другой стране" в 2023 году
top_ans_23 = [] # рейтинг ответов за 2023 год
for k, v in sorted(result_d.items(), key = lambda x: x[1][-1], reverse = True):
    top_ans_23.append(k)
place_23 = top_ans_23.index('За границей/в другой стране') + 1 # позиция, которую занимает ответ "За границей/в другой стране" в 2023 году
```

По данным анализа ответ "За границей, в другой стране" в 2023 году занимает 7 место (последним идет ответ "Затрудняюсь ответить", так что можно заключить, что вариант "За границей, в другой стране" является последним среди содержательных) в топе ответов респондентов. 
На основании этих данных можно сказать, что россияне предпочитают отдыхать внутри страны. И какие-либо внешние воздействия не оказывают значительного влияния на выбор места проведения отпуска, так как данные до пандемии и начала СВО не сильно отличаются от данных, полученных в 2023 году.

```python
# ищем процентное изменение по ответу "За границей/в другой стране"
diff = (result_d['За границей/в другой стране'][-1] - result_d['За границей/в другой стране'][0]) / result_d['За границей/в другой стране'][0] * 100
```

Посчитав процентное изменение количества отдохнувших за границей россиян, мы выяснили, что эта доля в значительной степени уменьшилась (на 62,5%). Скорее всего это связано, как уже отмечалось, с падением курса рубля и подорожанием путевок, а также рядом ограничений на въезд в другие страны. Интересно что, в 2021 году доля людей, отдыхавших за границей, составляла 13% и достигла максимума за весь период. Примечательно, что в 2020 году за границей было 0% опрошенных (вероятно из-за ковидных ограничений), что объясняет такой “бум” в следующем 2021 году, когда ограничения были уже сняты.

```python
# запишем все в итоговый файл
with open('result.txt', mode = 'w', encoding = 'utf-8') as outfile:
    print(f'В среднем за 5 лет (с 2019 по 2023 гг.) отпуск за границей провели {aver}% россиян.', file = outfile)
    print(' ', file = outfile)
    print(f'В 2019 году отпуск за границей был на {place_19} месте по популярности у россиян.', file = outfile)
    print(f'Топ ответов, где россияне провели лето 2019 года:', file = outfile)
    for i in range(len(top_ans_19)):
        print(f'{i + 1}. {top_ans_19[i].replace("/", ", ")}', file = outfile)
    print(' ', file = outfile)
    print(f'В 2023 году отпуск за границей был на {place_23} месте по популярности у россиян.', file = outfile)
    print(f'Топ ответов, где россияне провели лето 2023 года:', file = outfile)
    for i in range(len(top_ans_23)):
        print(f'{i + 1}. {top_ans_23[i].replace("/", ", ")}', file = outfile)
    print(' ', file = outfile)
    print(f'За 5 лет (с 2019 по 2023 г.г.) россияне стали на {str(diff).strip("-")}% реже ездить за границу.', file = outfile)
    print(' ', file = outfile)
```

Завершив обработку и анализ первого файла с данными, мы записали все результаты в новый файл вывода, в котором отразили все найденные описательные статистики, а также составили рейтинги ответов на основании того, где россияне провели отпуск в 2019 и 2023 гг.

```python
# обработка 2-ого файла

countries2019 = {} # страна: число выездных поездок из России в 2019 году
countries2023 = {} # страна: число выездных поездок из России в 2023 году
all = {2019: 0, 2023: 0}  # общее число выездных поездок из России в 2019 и 2023 годах

with open('tourism2.csv', mode = 'r', encoding = 'utf-8') as infile2:
    for line in infile2:
        line = line.strip().split(',')
        country, num1, num2 = line
        # заменяем пропущенные значения во втором столбце на 0
        if num1 == '-' or num1 == '':
            num1 = 0
        else:
            num1 = int(num1.replace(' ', '')) # преобразовываем строки в числа
        # делаем то же самое со следующим столбцом
        if num2 == '-' or num2 == '':
            num2 = 0
        else:
            num2 = int(num2.replace(' ', '')) # преобразовываем строки в числа
        # заполняем словарь all 
        if country == 'Всего':
            all[2019] = num1
            all[2023] = num2
        # заполняем словари countries2019 и countries2023
        elif country != 'Страна' and country[0].isupper():
            countries2019[country] = num1
            countries2023[country] = num2
```

```python
# найдем, во сколько раз меньше стало выездных поездок в 2023 году по сравнению с 2019 
difference = round(all[2019] / all[2023], 2)
```

Посчитав, во сколько раз меньше стало выездных туристских поездок в 2023 году по сравнению с 2019 мы можем более наглядно увидеть, как сильно падение курса рубля, подорожание путевок и введение ограничений на въезд для россиян в ряд стран повлияли на их отдых за границей.

```python
for k, v in countries2019.items():
    if v == min(countries2019.values()):
        unp_2019 = k # страна, в которую меньше всего выезжали из России в 2019 году
        perc_unp_2019 = round(v / all[2019] * 100, 2) # процентная доля поездок, совершенных в эту страну
    if v == max(countries2019.values()):
        pop_2019 = k # страна, в которую больше всего выезжали из России в 2019 году
        perc_pop_2019 = round(v / all[2019] * 100, 2) # процентная доля поездок, совершенных в эту страну
```

Вычислив минимум по значениям словаря, мы увидели, что меньше всего поездок из России в 2019 году было совершено в Гонконг. Вероятно, он не пользовался особым спросом у российских туристов, поскольку является специальным административным регионом Китая и представляет собой мегаполис, т.е., проще говоря, не туристический район. Кроме того, стоит отметить, что без визы в Гонконге можно находиться не более 14 дней, а так же, тот факт, что в Гонконг россияне вполне могли попасть из Китая, поэтому непосредственно в Гонконг было совершено мало поездок.

Далее мы вычислили процентную долю поездок, совершенных в Гонконг, чтобы более наглядно продемонстрировать, как мало туда выезжали россияне на фоне всех выездных туристских поездок за 2019 год. Процентная доля поездок в Гонконг составила всего 0,04% (хотя 17 тысяч поездок за год вполне могут казаться большим количеством).

Вычислив максимум по значениям словаря, мы увидели, что больше всего поездок в 2019 году было совершено в Турцию. Это можно объяснить отсутствием необходимости делать визу для въезда в данную страну, относительно невысокими ценами на отдых, наличием теплого климата, моря и относительно недолгим временем полета, что как раз и привлекает россиян, которые любят "ленивый" отдых.

Далее мы вычислили процентную долю поездок, совершенных в Турцию, чтобы продемонстрировать популярность Турции среди российских туристов более наглядно. Эта доля составила 14,9%.

```python
unp_2023 = [] # страны, в которые никто не выезжал из России в связи с ограничениями (НО россияне могли попасть в них через другие страны путем пересадок)
for k, v in countries2023.items():
    if v == min(countries2023.values()): # min = 0
        unp_2023.append(k)
    if v == max(countries2023.values()):
        pop_2023 = k # страна, в которую больше всего выезжали из России в 2023 году
        perc_pop_2023 = round(v / all[2023] * 100, 2) # процентная доля поездок, совершенных в эту страну
```

Так как при обработке второго файла все пустые ячейки мы заменили на 0, то в словаре с данными за 2023 год было несколько стран с минимальным значением (min = 0). Таким образом, список "unp_2023" - это список стран, в которые не было совершено выездных поездок из России в 2023 году. Среди них: страны, которые перестали выдавать визы россиянам; страны, закрывшие воздушное пространство для российских самолетов и т.п. Однако стоит отметить, что россияне могли попасть в вышеперечисленные страны путем пересадок, например, через Стамбул (Турция).

Далее мы вычислили максимум по значениям словаря с поездками за 2023 год и увидели, что больше всего поездок в 2023 году было совершено в Абхазию, что можно объяснить опять же отсутствием необходимости делать визу, относительно низкими ценами на отдых, наличием теплого климата, моря и близким расположением с Россией. Кроме того, стоит отметить, что для граждан России, осуществляющих въезд в Абхазию, даже не нужен загранпаспорт. Необходимо иметь при себе лишь паспорт гражданина РФ.

После этого мы снова вычислили процентную долю поездок, чтобы более наглядно продемонстрировать популярность Абхазии как места отдыха российских туристов. Она составила 22,47%.

```python
# добавим в итоговый файл результаты по обработке второго файла
with open('result.txt', mode = 'a', encoding = 'utf-8') as outfile:
    print(f'По сравнению с 2019 годом выездных туристских поездок в 2023 году стало в {difference} раз меньше.', file = outfile)
    print(' ', file = outfile)
    print(f'В 2019 году меньше всего выездных поездок из России было совершено в страну: {unp_2019} ({perc_unp_2019}%).', file = outfile)
    print(f'В 2019 году больше всего выездных поездок из России было совершено в страну: {pop_2019} ({perc_pop_2019}%).', file = outfile)
    print(f'В 2023 году не было совершено выездных поездок из России в следующие страны: {", ".join(unp_2023)}.', file = outfile)
    print(f'В 2019 году больше всего выездных поездок из России было совершено в страну: {pop_2023} ({perc_pop_2023}%).', file = outfile)
    print(' ', file = outfile)
    print(f'Топ-5 наиболее популярных стран для туризма среди россиян в 2019 году: ', file = outfile)
    # создаем рейтинг наиболее популярных стран за 2019 год
    rate1 = []
    for k, v in sorted(countries2019.items(), key = lambda x: x[1], reverse = True):
        rate1.append((k, v)) 
    rate1 = rate1[0:5] # оставляем топ-5
    for i in range(len(rate1)):
        print(f'{i + 1}. {rate1[i][0]} - {rate1[i][1]} тыс. ({round(rate1[i][1] / all[2019] * 100, 2)}%)', file = outfile)
    print(' ', file = outfile)    
    print(f'Топ-5 наиболее популярных стран для туризма среди россиян в 2023 году: ', file = outfile)
    # создаем рейтинг наиболее популярных стран за 2023 год
    rate2 = []
    for k, v in sorted(countries2023.items(), key = lambda x: x[1], reverse = True):
        rate2.append((k, v))
    rate2 = rate2[0:5]
    for i in range(len(rate2)):
        print(f'{i + 1}. {rate2[i][0]} - {rate2[i][1]} тыс. ({round(rate2[i][1] / all[2023] * 100, 2)}%)', file = outfile)
```

Завершив обработку и анализ второго файла с данными, мы добавили все результаты в итоговый файл вывода, в котором отразили все найденные описательные статистики, а также составили рейтинги наиболее популярных стран для туризма среди россиян в 2019 и 2023 гг. Примечательно, что процентные доли выездных поездок в Абхазию и Турцию в 2023 году стали значительно выше, чем в 2019, что, как уже отмечалось ранее, произошло из-за введения ограничений на въезд для россиян во многие другие страны.
